# Phase 3 — Matplotlib Foundations

Matplotlib is the foundation of all Python visualization. Seaborn, pandas `.plot()`, and even parts of scikit-learn are built on top of it. Learn this first.

**Core concept:** Everything in matplotlib lives inside a `Figure` (the whole image) containing one or more `Axes` (each individual plot).

---

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

np.random.seed(42)

# Use inline backend for notebooks
%matplotlib inline
```

---
## 1. Figure and Axes — The Core Architecture

There are two styles in matplotlib:
- **Pyplot style** (implicit): `plt.plot()`, `plt.show()` — quick and convenient for single plots
- **Object-oriented style** (explicit): `fig, ax = plt.subplots()` — recommended for all real work

```python
x = np.linspace(0, 2 * np.pi, 100)

# --- Pyplot style (quick) ---
plt.figure(figsize=(8, 3))
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")
plt.title("Pyplot style")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
# --- Object-oriented style (recommended) ---
fig, ax = plt.subplots(figsize=(8, 3))

ax.plot(x, np.sin(x), color="steelblue", linewidth=2, linestyle="-", label="sin(x)")
ax.plot(x, np.cos(x), color="coral", linewidth=2, linestyle="--", label="cos(x)")

ax.set_title("Object-Oriented style", fontsize=14)
ax.set_xlabel("x (radians)")
ax.set_ylabel("y")
ax.legend(loc="upper right")
ax.axhline(0, color="black", linewidth=0.8, linestyle=":")

plt.tight_layout()
plt.show()
```

---
## 2. Subplots — Multiple Plots in One Figure

```python
# 2 rows × 3 columns grid
fig, axes = plt.subplots(2, 3, figsize=(14, 8))

data = np.random.randn(200)

# Row 0: Line, Scatter, Bar
axes[0, 0].plot(np.cumsum(np.random.randn(100)), color="steelblue")
axes[0, 0].set_title("Line Plot (Cumulative Sum)")

axes[0, 1].scatter(
    np.random.randn(100), np.random.randn(100), alpha=0.6, color="coral", s=40
)
axes[0, 1].set_title("Scatter Plot")

categories = ["A", "B", "C", "D", "E"]
values = [23, 45, 12, 67, 34]
axes[0, 2].bar(categories, values, color="steelblue", edgecolor="white")
axes[0, 2].set_title("Bar Chart")

# Row 1: Histogram, Box plot, Pie
axes[1, 0].hist(data, bins=25, color="steelblue", edgecolor="white")
axes[1, 0].set_title("Histogram")

groups = [np.random.normal(loc, 1, 50) for loc in [0, 2, 4]]
axes[1, 1].boxplot(groups, labels=["Group A", "Group B", "Group C"])
axes[1, 1].set_title("Box Plot")

axes[1, 2].pie(
    values,
    labels=categories,
    autopct="%1.1f%%",
    colors=["steelblue", "coral", "green", "gold", "purple"],
)
axes[1, 2].set_title("Pie Chart")

plt.suptitle("Matplotlib Chart Gallery", fontsize=16, y=1.01)
plt.tight_layout()
plt.show()
```

## 3. Statistical Plots — Applying to Stats Concepts

Visualize every stats concept you've learned.

```python
from scipy import stats

# Histogram + Normal distribution overlay (check normality assumption)
sample = np.random.normal(loc=70, scale=12, size=300)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Histogram with density curve
axes[0].hist(
    sample,
    bins=25,
    density=True,
    color="steelblue",
    edgecolor="white",
    alpha=0.7,
    label="Data",
)
x_fit = np.linspace(sample.min(), sample.max(), 200)
axes[0].plot(
    x_fit,
    stats.norm.pdf(x_fit, sample.mean(), sample.std()),
    "r-",
    lw=2,
    label="Normal fit",
)
axes[0].axvline(
    sample.mean(), color="red", linestyle="--", label=f"Mean={sample.mean():.1f}"
)
axes[0].axvline(
    np.median(sample),
    color="green",
    linestyle="--",
    label=f"Median={np.median(sample):.1f}",
)
axes[0].set_title("Distribution with Normal Overlay")
axes[0].set_xlabel("Score")
axes[0].legend(fontsize=9)

# Box plot with annotations
q1, q2, q3 = np.percentile(sample, [25, 50, 75])
iqr = q3 - q1

bp = axes[1].boxplot(sample, vert=True, patch_artist=True)
bp["boxes"][0].set_facecolor("lightblue")
axes[1].set_title("Box Plot — IQR & Outliers")
axes[1].set_ylabel("Score")
axes[1].text(1.15, q1, f"Q1={q1:.1f}", va="center", fontsize=9)
axes[1].text(1.15, q2, f"Q2={q2:.1f}", va="center", fontsize=9)
axes[1].text(1.15, q3, f"Q3={q3:.1f}", va="center", fontsize=9)
axes[1].text(
    1.15, (q1 + q3) / 2, f"IQR={iqr:.1f}", va="center", color="steelblue", fontsize=9
)

plt.tight_layout()
plt.show()
```

```python
# Scatter matrix — relationships between variables (preview of EDA)
n = 100
age = np.random.normal(35, 10, n)
income = 20000 + age * 1500 + np.random.normal(0, 8000, n)
score = 40 + income / 2000 + np.random.normal(0, 5, n)

variables = {"Age": age, "Income": income, "Score": score}
names = list(variables.keys())
values_list = list(variables.values())

fig, axes = plt.subplots(3, 3, figsize=(10, 9))

for i, (name_i, vals_i) in enumerate(variables.items()):
    for j, (name_j, vals_j) in enumerate(variables.items()):
        ax = axes[i, j]
        if i == j:
            ax.hist(vals_i, bins=20, color="steelblue", edgecolor="white")
            ax.set_title(name_i, fontsize=10)
        else:
            r, _ = stats.pearsonr(vals_j, vals_i)
            ax.scatter(vals_j, vals_i, s=15, alpha=0.5, color="coral")
            ax.set_title(f"r={r:.2f}", fontsize=9)
        if j == 0:
            ax.set_ylabel(name_i, fontsize=9)
        if i == 2:
            ax.set_xlabel(name_j, fontsize=9)

plt.suptitle("Scatter Matrix — Pairwise Relationships", fontsize=13, y=1.01)
plt.tight_layout()
plt.show()
```

## 4. Styling and Customization

```python
# Available styles
print("Available styles:", plt.style.available[:10], "...")
```

```python
# Publication-quality plot with all common customizations
with plt.style.context("seaborn-v0_8-whitegrid"):
    fig, ax = plt.subplots(figsize=(9, 5))

    months = [
        "Jan",
        "Feb",
        "Mar",
        "Apr",
        "May",
        "Jun",
        "Jul",
        "Aug",
        "Sep",
        "Oct",
        "Nov",
        "Dec",
    ]
    sales_2024 = [120, 135, 148, 162, 175, 190, 185, 200, 195, 210, 230, 260]
    sales_2025 = [130, 145, 160, 178, 192, 205, 200, 218, 210, 228, 248, 280]

    x = np.arange(len(months))
    width = 0.35

    bars1 = ax.bar(
        x - width / 2,
        sales_2024,
        width,
        label="2024",
        color="steelblue",
        alpha=0.85,
        edgecolor="white",
    )
    bars2 = ax.bar(
        x + width / 2,
        sales_2025,
        width,
        label="2025",
        color="coral",
        alpha=0.85,
        edgecolor="white",
    )

    ax.set_xlabel("Month", fontsize=12)
    ax.set_ylabel("Sales ($1000s)", fontsize=12)
    ax.set_title(
        "Monthly Sales Comparison: 2024 vs 2025", fontsize=14, fontweight="bold"
    )
    ax.set_xticks(x)
    ax.set_xticklabels(months)
    ax.legend(fontsize=11)
    ax.set_ylim(0, 320)

    # Add value labels on top of bars
    for bar in bars2:
        height = bar.get_height()
        ax.annotate(
            f"{height}",
            xy=(bar.get_x() + bar.get_width() / 2, height),
            xytext=(0, 3),
            textcoords="offset points",
            ha="center",
            va="bottom",
            fontsize=8,
            color="gray",
        )

    plt.tight_layout()
    plt.show()
```

## 5. Saving Figures

```python
# Save a figure to disk
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot([1, 2, 3, 4], [1, 4, 9, 16], "o-", color="steelblue")
ax.set_title("Saved Figure Example")

# dpi=150 is good for reports; dpi=300 for publication
fig.savefig("example_plot.png", dpi=150, bbox_inches="tight")
print("Figure saved as 'example_plot.png'")
plt.show()
```

## Summary — Matplotlib Essentials

| Task | Code |
|------|------|
| Create figure | `fig, ax = plt.subplots(figsize=(w,h))` |
| Multiple subplots | `fig, axes = plt.subplots(rows, cols)` |
| Line plot | `ax.plot(x, y, color, linestyle, lw, label)` |
| Scatter plot | `ax.scatter(x, y, s, alpha, c)` |
| Histogram | `ax.hist(data, bins, density, color)` |
| Bar chart | `ax.bar(x, height, width, color)` |
| Box plot | `ax.boxplot(data, labels, patch_artist)` |
| Add labels | `ax.set_title()`, `ax.set_xlabel()`, `ax.set_ylabel()` |
| Add legend | `ax.legend()` — requires `label=` in plots |
| Save figure | `fig.savefig('file.png', dpi=150, bbox_inches='tight')` |
