### Estimate

**Inferential Statistics** me "**estimate**" ka matlab hota hai kisi **population parameter** (jaise population ka mean ya proportion) ko ek **sample data ke basis pe approximate karna**.

Jab bhi hum kisi bade group (**population**) ke baare me kuch conclude karte hain sample data ke use se, to usse "**estimation**" kehte hain.

Types of Estimates in Inferential Statistics:
1. **Point Estimate** - This house costs you around $10,000.
2. **Interval Estimate** - This house costs you between $8,000 and $12,000.

**Point Estimate:**

* Definition: Point estimate ek **single value** hoti hai jo **sample** data se kisi **population** parameter ka **best** guess hoti hai.
* Example: Agar aapke sample ka mean 50 hai, toh aapka **point estimate population mean** ke liye **50** hoga.

**Interval Estimate:**

* Definition: Interval estimate ek **range** hoti hai jo **sample** data se kisi **population** parameter ka best guess hoti hai.
* Example: Agar aapke sample ka mean 50 hai, toh aapka interval estimate population mean ke liye **$40 to $60** hoga.

**Interval Estimate:**

* **Definition**: Interval estimate ek **range** provide karta hai jismein **population parameter** hone ki **probability** hoti hai. Yeh zyada reliable hota hai kyunki yeh ek **range** provide karta hai, na ki ek **single value**.

* **Example**: Agar aapke **sample** data se aapka interval estimate **45 se 55 hai**, toh aap keh sakte hain ki population mean **45 se 55** ke beech ho sakta hai with a **certain level of confidence**.

**Confidence Interval (CI)**

ek **range** hota hai jo **estimate** karta hai ki kisi **population** parameter ki value kis **range** ke andar fall kar sakti hai, aur yeh ek confidence level ke sath hota hai.

Confidence level yeh batata hai ki agar aap multiple samples lenge aur unke confidence intervals calculate karenge, toh usme se kitne intervals population parameter ko cover karenge.

**For Example:**

Agar aapke **sample** mean ke liye **95% confidence interval** **45 se 55** hai, iska matlab hai ki aapke **population** mean ka 95% chance hai ki yeh **range (45 se 55)** ke andar hoga.

**Hypothesis Testing**

inferential statistics mein ek **fundamental method** hai jo help karta hai, data ke basis pe **decisions** lene aur **conclusions** draw karne mein. Yeh **process** decide karta hai ki **sample** data ke **basis** pe **population** ke baare mein koi **specific claim ya hypothesis** true hai ya nahi.

**Hypothesis Testing Kya Hai?**

Hypothesis Testing ek **systematic method** hai jisme hum ek **hypothesis (assumption) ko test** karte hain aur decide karte hain ki **sample data** us hypothesis ko **support** karta hai ya nahi. Yeh statistical inference ka ek part hai jo ki probability theory pe based hota hai.

**Hypothesis Testing Mechanism**

1. **Null Hypothesis (HO)**

The **assumptions** you are beginning with

2. **Alternative Hypothesis (H1)**

**Opposite** of Null Hypothesis

3. **Experiments**

Use **Statistical methods** (such as t-tests, z-tests, ANOVA, chi-square tests,etc) to analyze the
data.

4. **Accept the null hypothesis or reject the null hypothesis**

It helps researchers and analysts assess the validity of claims, evaluate the effectiveness interventions and **draw** **meaningful** **conclusions** about **population**.

**p-Value Test**

p-Value Test inferential statistics mein ek **key concept** hai jo **hypothesis** testing ke process ka essential part hota hai.

Yeh ek measure hai jo batata hai ki **sample** data, **null hypothesis** ke saath **kitna consistent** hai. Specifically, **p-value** se yeh assess kiya jata hai ki null hypothesis ko reject karne ke liye aapke sample data ke basis pe kitni evidence hai.

**p-Value Kya Hai?**

p-Value ek **probability** hai jo aapke sample data ya usse extreme results aane ki probability ko represent karta hai, assuming ki null hypothesis sach hai.

Yeh batata hai ki aapke observed results (ya aur extreme results) ko observe karna kitna unlikely hai agar null hypothesis sach hoti.

**T Test**

In statistics, a t-test is a **hypothesis test** used to **determine** if there is a **significant difference** **between the means of two groups**. It's especially **useful** when **dealing** with **small sample sizes** or when the **population** standard deviation is **unknown**.

```python
import numpy as np
from scipy import stats

# Step 1: Create two random sample datasets

# Suppose these are test scores from two groups (e.g., group A and group B)

group_A = np.random.normal(75, 10, 30)  # Group A with mean 75, std 10, and 30 samples
group_B = np.random.normal(80, 12, 30)  # Group B with mean 80, std 12, and 30 samples

# Step 2: Perform an independent t-test
t_statistic, p_value = stats.ttest_ind(group_A, group_B)

# Step 3: Display the p-value and t-statistic
print(f"T-statistic: {t_statistic}")
print(f"p-value: {p_value}")
print("\n")

# Step 4: Interpretation based on p-value
alpha = 0.05  # Significance Level(95% confidence level)
# 5% is the most common default in science, medicine, and engineering.
# 0.10 - More lenient - easier to call a result "significant"
# 0.05 - Standard for most social science research
# 0.01 - More stringent(stricter), harder to reject the null hypothesis

if p_value <= alpha:
    print(
        f"Reject the null hypothesis\n (Significant difference between groups)\n p-value: {p_value} is less than alpha: {alpha}"
    )
else:
    print(
        f"Fail to reject the null hypothesis\n (No significant difference between groups)\n p-value: {p_value} is greater than alpha: {alpha}"
    )
```

**Simple analogy**

* p-value = strength of evidence against “no difference”
* alpha = how strong evidence you require before acting

Alpha is your **threshold**, not something scipy calculates. The **t-test** only gives you the **p-value**; you compare it to **alpha** to make the final call.

**Z - Test**

Z-Value Test (Z-Test) inferential statistics ka ek **method** hota hai, jo tab use hota hai jab aapko test karna ho ki **sample ka mean** **population mean** ke **barabar** hai ya nahi.

Z-test ko tab use kiya jata hai jab **sample size bada** ho (usually n>30n > 30n>30) ya jab population ka **standard deviation sigma pata** ho.

**Example:-**

The **average height** of **all** the residents in a city is **168 cm**, with a **standard deviation = 3.9**.

A doctor **believes** the **mean** to be different. He measured the height of 36 **individuals** and found the **average height to be 169.5 cm**.

1. State null and alternate hypothesis
2. At a 95% confidence level, is there enough evidence to reject the null hypothesis

```python
import numpy as np
from scipy import stats

# Given information
pop_mean = 168  # population mean height (cm)
pop_std = 3.9  # population standard deviation (cm)
sample_size = 36
sample_mean = 169.5  # average height from the doctor's sample (cm)

# 1. State null and alternate hypothesis
print("Null Hypothesis (H0): The population mean height is 168 cm (μ = 168)")
print(
    "Alternate Hypothesis (H1): The population mean height is different from 168 cm (μ ≠ 168)"
)
print()

# Z-test: population std is known and sample size is large (n = 36)
z_statistic = (sample_mean - pop_mean) / (pop_std / np.sqrt(sample_size))
p_value = 2 * (1 - stats.norm.cdf(abs(z_statistic)))  # two-tailed test

print(f"Z-statistic: {z_statistic:.4f}")
print(f"p-value: {p_value:.4f}")
print()

# 2. Decision at 95% confidence level (alpha = 0.05)
alpha = 0.05

if p_value <= alpha:
    print(
        f"Reject the null hypothesis.\n"
        f"There is enough evidence at the 95% confidence level to say the mean height is different from 168 cm.\n"
        f"(p-value {p_value:.4f} ≤ alpha {alpha})"
    )
else:
    print(
        f"Fail to reject the null hypothesis.\n"
        f"There is not enough evidence at the 95% confidence level to say the mean height is different from 168 cm.\n"
        f"(p-value {p_value:.4f} > alpha {alpha})"
    )
```

**Q: Why did we choose a one-sample z-test?**

A **one-sample z-test** fits this problem because of what we are comparing and what we already know.

**1. One sample, not two groups**

The doctor measured **one group** of 36 people and asks: is their average (169.5 cm) different from the **known city average** (168 cm)?

That is **one sample vs a claimed population mean** — not Group A vs Group B. So it is **one-sample**, not two-sample (like the t-test example with `group_A` and `group_B`).

**2. Population standard deviation is known**

The problem gives **σ = 3.9** for the whole city. When σ is known, the formula uses it:

z = (x̄ − μ) / (σ / √n)

A **t-test** is used when σ is **unknown** and we estimate spread from the sample using **s** (sample std). Here σ is given, so z is the right choice.

**3. Large enough sample (n = 36)**

With **n = 36** (> 30), the sampling distribution of the mean is close to normal, which supports a z-test.

**Quick comparison**

| Question | Test to use |
|----------|-------------|
| Is this sample's mean different from a known μ, with **known σ**? | **One-sample z-test** ← height example |
| Is this sample's mean different from a known μ, with **unknown σ**? | One-sample t-test |
| Are **two groups'** means different? | Two-sample t-test ← earlier t-test cell |

**Summary:** one sample, known population mean (168), known σ (3.9) → **one-sample z-test**.

A company claims that the **average weight** of their **product** is **500 grams**. We want to test if this **claim is true** by **collecting a sample of 30 products**.

The **sample mean is 505 grams**, and the **population standard deviation is 10 grams**. What **test** should we conduct to see if the **difference in the mean** is **significant** or not?

```python
import numpy as np
from scipy import stats

# ---------------------------------------------------------------------------
# Step 1: Understand the problem and choose the right test
# ---------------------------------------------------------------------------
# We have ONE sample (30 products) and want to compare its mean to a CLAIMED
# population mean (500 g). Population standard deviation (σ = 10) is KNOWN.
# → Use a one-sample z-test (same logic as the height example).

claimed_mean = 500  # μ₀: what the company claims (grams)
pop_std = 10  # σ: known population standard deviation (grams)
sample_size = 30  # n: number of products we measured
sample_mean = 505  # x̄: average weight from our sample (grams)

print("Step 1 — Test choice: one-sample z-test")
print("  Reason: one sample, known σ, comparing sample mean to claimed μ₀")
print()

# ---------------------------------------------------------------------------
# Step 2: State the null and alternative hypotheses
# ---------------------------------------------------------------------------
# H₀: The company's claim is true → true mean weight = 500 g
# H₁: The claim is not true → true mean weight ≠ 500 g (two-tailed test)

print("Step 2 — Hypotheses")
print("  H₀ (null):        μ = 500 g  (company claim is correct)")
print("  H₁ (alternative): μ ≠ 500 g  (true mean is different from 500 g)")
print()

# ---------------------------------------------------------------------------
# Step 3: Calculate the z-statistic
# ---------------------------------------------------------------------------
# Formula: z = (x̄ − μ₀) / (σ / √n)
#   x̄ − μ₀  → how far our sample mean is from the claimed mean
#   σ / √n   → standard error (typical spread of sample means)
#   z        → how many "standard errors" away we are from the claim

standard_error = pop_std / np.sqrt(sample_size)
z_statistic = (sample_mean - claimed_mean) / standard_error

print("Step 3 — Z-statistic")
print(f"  Standard error (σ/√n): {standard_error:.4f} g")
print(f"  z = (x̄ − μ₀) / SE = ({sample_mean} − {claimed_mean}) / {standard_error:.4f}")
print(f"  z-statistic: {z_statistic:.4f}")
print()

# ---------------------------------------------------------------------------
# Step 4: Calculate the p-value
# ---------------------------------------------------------------------------
# p-value = probability of seeing a sample mean this extreme (or more)
#           IF the null hypothesis (μ = 500) were actually true.
# Two-tailed: we check both tails (mean could be higher OR lower than 500).

p_value = 2 * (1 - stats.norm.cdf(abs(z_statistic)))

print("Step 4 — p-value")
print(f"  p-value (two-tailed): {p_value:.4f}")
print("  Interpretation: if the true mean were 500 g, there is a")
print(
    f"  {p_value * 100:.2f}% chance of observing a sample mean at least this far from 500 g."
)
print()

# ---------------------------------------------------------------------------
# Step 5: Compare p-value to alpha and make a decision
# ---------------------------------------------------------------------------
# alpha = 0.05 is the standard significance level (95% confidence).
# If p-value ≤ alpha → reject H₀ (difference is statistically significant)
# If p-value > alpha  → fail to reject H₀ (not enough evidence)

alpha = 0.05

print("Step 5 — Decision (α = 0.05, 95% confidence level)")
if p_value <= alpha:
    print(f"  Reject H₀ (p-value {p_value:.4f} ≤ α {alpha})")
    print("  Conclusion: The difference IS statistically significant.")
    print("  The sample evidence suggests the true mean weight is NOT 500 g.")
else:
    print(f"  Fail to reject H₀ (p-value {p_value:.4f} > α {alpha})")
    print("  Conclusion: The difference is NOT statistically significant.")
    print("  We do not have enough evidence to dispute the company's claim.")
```
